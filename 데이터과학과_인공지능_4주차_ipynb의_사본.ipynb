## 2. 변수와 자료형

### 변수: 값에 이름표 붙이기

**변수**는 값에 붙이는 이름표입니다. `=` 기호로 붙입니다.

> 주의: 여기서 `=` 는 "같다"가 아니라 **"오른쪽 값을 왼쪽 이름에 넣어라"** 라는 뜻입니다. 수학의 등호와 다릅니다.

````python
이름 = "김고려"
나이 = 20
학점 = 3.8

print(이름)
print(나이)
print(학점)
````

Python은 한글 변수명도 됩니다. 다만 실무에서는 영어를 쓰는 게 일반적이고, 오류 메시지나 AI에게 물어볼 때도 영어 쪽이 편합니다. 이 수업에서는 둘 다 허용합니다.

변수는 **덮어쓸 수** 있습니다. 나중에 넣은 값이 출력됩니다.

````python
점수 = 80
print(점수)

점수 = 95
print(점수)
````

### 자료형: 값의 종류

값에는 종류가 있습니다. 오늘은 네 가지만 알면 됩니다.

| 자료형 | 영어 이름 | 예시 | 설명 |
|---|---|---|---|
| 정수 | `int` | `20`, `-3`, `0` | 소수점 없는 숫자 |
| 실수 | `float` | `3.8`, `-0.5` | 소수점 있는 숫자 |
| 문자열 | `str` | `"김고려"`, `"20"` | 따옴표로 감싼 글자 |
| 불리언 | `bool` | `True`, `False` | 참/거짓 |

자료형이 궁금하면 `type()` 으로 확인합니다.

````python
print(type(20))
print(type(3.8))
print(type("김고려"))
print(type(True))
````

### 왜 자료형을 알아야 하나

**같은 `20` 이라도 `20` 과 `"20"` 은 다른 값**이기 때문입니다. 아래 셀을 실행해 보세요.

````python
print(20 + 5)
print("20" + "5")
````

`"20" + "5"` 가 `25` 가 아니라 `205` 로 나왔습니다.

이건 다음 주부터 자주 겪게 될 문제입니다. 엑셀 파일에서 숫자를 읽어왔는데 Python이 그걸 글자로 인식해서 계산이 안 되는 상황이 아주 흔합니다. 그때 `type()` 으로 확인하면 됩니다.

글자와 숫자를 섞어서 출력하고 싶을 때는 **f-문자열**을 씁니다. 따옴표 앞에 `f` 를 붙이고, 변수를 `{ }` 로 감쌉니다.

````python
이름 = "김고려"
나이 = 20

print(f"{이름}님은 {나이}살입니다.")
print(f"내년에는 {나이 + 1}살이 됩니다.")
````

## 3. 리스트: 값을 여러 개 담기

변수 하나에 값 하나씩 담으면 불편합니다. 학생 30명의 점수를 담으려면 변수를 30개 만들어야 할까요?

**리스트**는 값을 여러 개 담는 상자입니다. 대괄호 `[ ]` 로 만들고, 쉼표로 구분합니다.

````python
점수들 = [88, 92, 75, 100, 64]
메뉴들 = ["아메리카노", "라떼", "녹차라떼"]

print(점수들)
print(메뉴들)
````

### 인덱싱: 몇 번째 값 꺼내기

리스트에서 값을 꺼낼 때는 대괄호에 **순서 번호(인덱스)** 를 씁니다.

**Python은 0부터 셉니다.** 첫 번째가 0번, 두 번째가 1번입니다.

```
점수들 = [ 88,  92,  75, 100,  64 ]
인덱스     0    1    2    3    4
뒤에서    -5   -4   -3   -2   -1
```

````python
점수들 = [88, 92, 75, 100, 64]

print(점수들[0])    # 첫 번째
print(점수들[2])    # 세 번째
print(점수들[-1])   # 마지막
````

### 슬라이싱: 여러 개를 한 번에 꺼내기

`[시작:끝]` 으로 구간을 꺼낼 수 있습니다.

**끝 번호는 포함되지 않습니다.** `[1:3]` 은 1번과 2번만 가져옵니다. 이것도 처음에는 헷갈립니다.

````python
점수들 = [88, 92, 75, 100, 64]

print(점수들[1:3])   # 1번부터 2번까지
print(점수들[:3])    # 처음부터 2번까지
print(점수들[2:])    # 2번부터 끝까지
````

### 리스트로 할 수 있는 것들

````python
점수들 = [88, 92, 75, 100, 64]

print(len(점수들))    # 몇 개인지
print(sum(점수들))    # 다 더하면
print(max(점수들))    # 가장 큰 값
print(min(점수들))    # 가장 작은 값
print(sum(점수들) / len(점수들))   # 평균
````

## 4. 딕셔너리: 이름표를 붙여서 담기

리스트는 "몇 번째"로 값을 꺼냈습니다. 그런데 `점수들[2]` 가 누구 점수인지는 알 수 없습니다.

**딕셔너리**는 값마다 **이름(키)** 을 붙여서 담습니다. 중괄호 `{ }` 를 쓰고, `키: 값` 형태로 적습니다.

````python
학생 = {"이름": "김고려", "학과": "경영학과", "학년": 1}

print(학생["이름"])
print(학생["학과"])
````

딕셔너리는 이번 주에 깊게 다루지 않습니다. 다만 **다음 주에 표 데이터를 만들 때 이 모양이 그대로 나오기 때문에** 미리 눈에 익혀두는 겁니다.

아래처럼 생긴 코드를 다음 주에 보게 됩니다. 지금은 "아, 딕셔너리구나" 정도만 알면 충분합니다.

````python
메뉴표 = {
    "메뉴": ["아메리카노", "라떼", "녹차라떼"],
    "가격": [4500, 5000, 5500]
}

print(메뉴표["메뉴"])
print(메뉴표["가격"])
````

## 5. 반복문: 같은 일을 여러 번

리스트 안의 값을 하나씩 꺼내서 똑같은 처리를 하고 싶을 때 **for 반복문**을 씁니다.

```python
for 변수 in 리스트:
    (들여쓰기) 반복할 내용
```

**콜론 `:` 과 들여쓰기가 반드시 필요합니다.** 들여쓰기가 빠지면 오류가 납니다. 코랩에서는 `Tab` 키를 누르면 됩니다.

````python
메뉴들 = ["아메리카노", "라떼", "녹차라떼"]

for 메뉴 in 메뉴들:
    print(메뉴)
````

`메뉴` 라는 이름은 여러분이 정하는 겁니다. `x` 라고 해도 되고 `item` 이라고 해도 됩니다. 다만 읽는 사람이 알아볼 수 있는 이름이 좋습니다.

조금 더 실용적인 예시를 보겠습니다.

````python
가격들 = [4500, 5000, 5500]

for 가격 in 가격들:
    print(f"{가격}원 → 세일가 {가격 * 0.9:.0f}원")
````

`{가격 * 0.9:.0f}` 에서 `:.0f` 는 **소수점 없이 보여달라**는 뜻입니다. 지금은 이런 게 있다는 정도만 알아두세요.

이번 주에 for문은 여기까지입니다. 다음 주부터 Pandas를 쓰면 **반복문 없이도 표 전체를 한 번에 계산**할 수 있어서, 사실 for문을 쓸 일이 많지는 않습니다. 하지만 코드를 읽으려면 알아야 합니다.

## 6. import 와 점 표기법 — 오늘의 핵심

이 절이 오늘 가장 중요합니다. **다음 주부터 계속 쓰게 될 문법**이기 때문입니다.

### import: 도구 상자 꺼내오기

Python 자체에는 기능이 많지 않습니다. 필요한 기능은 **도구 상자(라이브러리)** 를 꺼내 와서 씁니다. 그 명령이 `import` 입니다.

````python
import math

print(math.sqrt(16))    # 제곱근
print(math.pi)          # 원주율
````

이름이 길면 별명을 붙입니다. `as` 를 씁니다.

다음 주에 여러분이 매번 치게 될 줄이 바로 이겁니다.

```python
import pandas as pd
```

"pandas 라는 도구 상자를 가져오되, 앞으로는 `pd` 라고 줄여 부르겠다"는 뜻입니다.

### 점 표기법: `무엇.무엇()`

위 코드에서 `math.sqrt(16)` 을 봤습니다. 이 점 `.` 이 무슨 뜻일까요?

**"왼쪽 것에게 오른쪽 일을 시켜라"** 라는 뜻입니다. `math.sqrt(16)` 은 "math 도구 상자야, 16의 제곱근 좀 구해줘"입니다.

그리고 점 뒤에 오는 것은 두 종류입니다.

| 형태 | 이름 | 뜻 | 예시 |
|---|---|---|---|
| `.무엇()` | 메서드 | **시키는 일** (괄호 있음) | `"abc".upper()` |
| `.무엇` | 속성 | **가지고 있는 정보** (괄호 없음) | `math.pi` |

괄호가 있으면 "해줘", 없으면 "알려줘"라고 기억하세요.

````python
인사 = "안녕하세요 여러분"

print(인사.upper())           # 대문자로 (영어만 해당)
print(인사.replace("여러분", "1학년"))   # 바꿔치기
print(인사.split())           # 띄어쓰기로 쪼개기
print(인사.count("안"))        # 몇 번 나오는지
````

**왜 이게 중요한가.**

다음 주부터 여러분은 이런 코드를 치게 됩니다.

```python
df.head()          # 표야, 앞부분 좀 보여줘      → 메서드
df.shape           # 표야, 크기가 어떻게 돼?    → 속성
df.isna().sum()    # 표야, 빈칸 찾아서 세어줘    → 메서드 두 번
```

이 구조를 모르면 그냥 주문을 외우게 됩니다. 알고 나면 **"표에게 무슨 일을 시키는 중"** 으로 읽힙니다.

`df.isna().sum()` 처럼 점이 연달아 나오는 건, **앞의 결과에 다시 일을 시키는 것**입니다. 왼쪽부터 차례로 읽으면 됩니다.

## 7. 오류 메시지 읽기 — 겁내지 않기

코드를 짜다 보면 빨간 글씨가 쏟아집니다. **정상입니다.**

**맨 아래 마지막 줄만 읽으세요.** 위쪽은 대부분 컴퓨터 내부 사정이라 안 읽어도 됩니다.

아래 셀들을 **일부러 실행해서** 오류를 내 보겠습니다.

````python
# 오류 1: 없는 변수를 불렀을 때
print(존재하지않는변수)
````

마지막 줄: `NameError: name '존재하지않는변수' is not defined`

→ **"그런 이름 없는데요"** 입니다. 오타를 냈거나, 그 변수를 만든 셀을 아직 실행하지 않은 경우입니다.

코랩에서 제일 흔한 원인은 **위쪽 셀을 실행하지 않고 건너뛴 것**입니다. 위에서부터 순서대로 실행했는지 확인하세요.

````python
# 오류 2: 글자와 숫자를 더했을 때
"가격: " + 4500
````

마지막 줄: `TypeError: can only concatenate str (not "int") to str`

→ **"글자에는 글자만 붙일 수 있어요"** 입니다. 앞에서 본 그 문제입니다. f-문자열을 쓰면 해결됩니다.

````python
# 오류 3: 리스트 범위를 벗어났을 때
점수들 = [88, 92, 75]
print(점수들[5])
````

마지막 줄: `IndexError: list index out of range`

→ **"그 번호는 리스트에 없어요"** 입니다. 3개짜리 리스트의 인덱스는 0, 1, 2 까지입니다.

### 자주 만나는 오류 정리

| 오류 이름 | 무슨 뜻인가 | 보통 원인 |
|---|---|---|
| `NameError` | 그런 이름 없음 | 오타, 위 셀을 실행 안 함 |
| `TypeError` | 자료형이 안 맞음 | 글자와 숫자를 섞음 |
| `IndexError` | 그 번호 없음 | 리스트 범위를 벗어남 |
| `KeyError` | 그 키 없음 | 딕셔너리에 없는 이름을 부름 |
| `SyntaxError` | 문법이 깨짐 | 괄호·따옴표·콜론 빠짐 |
| `IndentationError` | 들여쓰기 문제 | for문 아래 들여쓰기 누락 |

`SyntaxError` 가 났는데 해당 줄이 멀쩡해 보인다면, **그 바로 윗줄**을 보세요. 괄호나 따옴표를 안 닫은 경우가 대부분입니다.
