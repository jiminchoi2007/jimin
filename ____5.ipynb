### 0.2. 괄호와 따옴표 치트시트 — 막히면 여기로 돌아오세요

지난주 가장 많이 나온 질문입니다. **"언제 `()`, 언제 `[]`, 언제 `{}` 를 쓰나요?"**

외우려 하지 말고, **각 기호가 하는 일이 다르다**는 것만 기억하세요.

| 기호 | 이름 | 하는 일 | 예시 |
|---|---|---|---|
| `( )` | 소괄호 | **일을 시킬 때** (함수·메서드 호출) | `print(...)`, `len(...)`, `df.head()` |
| `[ ]` | 대괄호 | **① 리스트 만들기 ② 꺼내기** | `[1, 2, 3]`, `점수들[0]`, `df["메뉴"]` |
| `{ }` | 중괄호 | **① 딕셔너리 만들기 ② f-문자열 안에서 변수 넣기** | `{"이름": "김고려"}`, `f"{나이}살"` |
| `" "` | 큰따옴표 | **글자라는 표시** | `"아메리카노"` |

한 문장으로 줄이면 이렇습니다.

> **소괄호는 "해줘", 대괄호는 "꺼내줘", 중괄호는 "여기에 끼워넣어"**

아래 셀을 실행하면서 눈으로 확인해 보세요.

````python
# ( ) — 일을 시킨다
print("안녕하세요")        # print에게 보여달라고 시킴
print(len("안녕하세요"))   # len에게 길이를 세달라고 시킴

# [ ] — 만들거나 꺼낸다
점수들 = [88, 92, 75]      # 리스트를 만든다
print(점수들[0])           # 0번째를 꺼낸다

# { } — 딕셔너리를 만들거나, f-문자열에 끼워넣는다
학생 = {"이름": "김고려"}   # 딕셔너리를 만든다
print(학생["이름"])        # 꺼낼 때는 [ ] 를 쓴다!

나이 = 20
print(f"저는 {나이}살입니다")   # f-문자열 안에 끼워넣는다
````

**헷갈리기 쉬운 지점 하나.** 딕셔너리는 `{ }` 로 **만들지만**, 값을 **꺼낼 때는 `[ ]`** 를 씁니다. 만드는 기호와 꺼내는 기호가 다릅니다.

```python
학생 = {"이름": "김고려"}    # 만들 때: 중괄호
학생["이름"]                 # 꺼낼 때: 대괄호
```

오늘 배울 `df["메뉴"]` 도 **꺼내는** 동작이라 대괄호입니다.

### 0.3. `NameError` — 셀 실행 순서 이야기

지난주 과제에서 **가장 많이 나온 오류**입니다.

```
NameError: name 'df' is not defined
```

코드를 아무리 봐도 멀쩡한데 이 오류가 납니다. 왜일까요?

**코랩은 위에서부터 순서대로 실행해야 합니다.** 셀을 하나 실행하면 그 결과(변수)가 컴퓨터 메모리에 저장되고, 아래 셀에서 그걸 꺼내 씁니다. 위 셀을 건너뛰면 **아래 셀이 쓸 재료가 없는 상태**가 됩니다.

직접 확인해 봅시다. 아래 두 셀 중 **두 번째 것을 먼저** 실행해 보세요.

````python
# ① 이 셀을 먼저 실행하면 아래 셀이 작동합니다
좋아하는메뉴 = "아메리카노"
````

````python
# ② 위 셀을 실행하지 않고 이것만 실행하면 NameError가 납니다
print(좋아하는메뉴)
````

`NameError` 가 났다면 위 셀을 실행하고 다시 해보세요. 이번엔 잘 됩니다.

### `NameError` 가 나면 할 일

**1. 위 셀들을 순서대로 실행했는지 확인**

셀 왼쪽에 `[1]`, `[2]` 처럼 번호가 붙습니다. 번호가 없으면 **아직 실행하지 않은 셀**입니다. 건너뛴 게 없는지 보세요.

**2. 오타 확인**

`메뉴` 와 `menu` 는 다른 이름입니다. `수량` 과 `수랑` 도 다릅니다. 지난주에 실제로 나온 사례입니다.

**3. 런타임이 끊겼는지 확인**

브라우저를 닫았다가 다시 열거나, 오래 방치하면 **지금까지 만든 변수가 전부 사라집니다.** 파일은 남아 있지만 메모리는 비워집니다.

이때는 당황하지 말고 **런타임 → 모두 실행** 을 누르세요. 처음부터 끝까지 다시 실행됩니다. 중간부터만 다시 하려면 **런타임 → 이전 셀 실행** 입니다.

> ### 오늘 학하다 막히면
>
> 1. 빨간 글씨의 **마지막 줄**을 읽는다
> 2. `NameError` 면 → 위 셀부터 다시 실행
> 3. 그래도 안 되면 → **런타임 → 모두 실행**
> 4. 여전히 안 되면 → 코드 전체와 오류 메시지를 통째로 복사해서 AI에게 묻거나 Q&A에 올린다

## 1. 데이터를 만나기

먼저 실습 데이터를 만듭니다. 아래 셀은 **그냥 실행**하세요.

````python
# 설정 셀 — 그냥 실행만 하세요 (이 코드는 몰라도 됩니다)
import random
random.seed(26)

메뉴정보 = {
    "아메리카노": ("커피", 4500), "카페라떼": ("커피", 5000),
    "바닐라라떼": ("커피", 5500), "콜드브루": ("커피", 5000),
    "녹차라떼": ("논커피", 5500), "초코라떼": ("논커피", 5500),
    "자몽에이드": ("논커피", 6000), "복숭아아이스티": ("논커피", 4500),
}
요일표 = ["월", "화", "수", "목", "금"]
좋은후기 = [
    "맛이 깔끔하고 좋았어요", "직원분이 친절하셨습니다", "자리가 넓어서 공부하기 좋아요",
    "가격 대비 양이 넉넉합니다", "향이 진하고 맛있어요", "분위기가 조용해서 좋네요",
]
보통후기 = [
    "무난한 맛입니다", "그냥 평범했어요", "나쁘지 않지만 특별하지도 않아요",
    "가격은 적당한 편", "다시 올지는 모르겠네요",
]
나쁜후기 = [
    "대기 시간이 너무 길었어요", "맛이 생각보다 밍밍합니다", "자리가 없어서 서서 마셨어요",
    "가격에 비해 아쉬웠습니다", "주문이 잘못 나왔어요", "대기가 길고 시끄러웠어요",
]

행들 = ["날짜,요일,매장,메뉴,분류,가격,수량,결제수단,평점,후기"]
for 주 in range(3):
    for i, 요일 in enumerate(요일표):
        날짜 = f"2026-03-{2 + 주 * 7 + i:02d}"
        for 매장 in ["안암점", "정문점"]:
            for 메뉴 in random.sample(list(메뉴정보), 3):
                분류, 가격 = 메뉴정보[메뉴]
                수량 = random.randint(5, 45)
                결제 = random.choice(["카드", "현금", "간편결제"])
                평점 = random.choices([1, 2, 3, 4, 5], weights=[3, 7, 20, 40, 30])[0]
                풀 = 좋은후기 if 평점 >= 4 else (보통후기 if 평점 == 3 else 나쁜후기)
                후기 = random.choice(풀)
                행들.append(f"{날짜},{요일},{매장},{메뉴},{분류},{가격},{수량},{결제},{평점},{후기}")

with open("cafe_sales_full.csv", "w", encoding="utf-8") as f:
    f.write("\n".join(행들))

print(f"cafe_sales_full.csv 생성 완료 — 총 {len(행들) - 1}행")
````

## 2. DataFrame과 Series — 표와 열

Pandas에는 두 가지 그릇이 있습니다.

| 이름 | 모양 | 비유 |
|---|---|---|
| **DataFrame** | 2차원 — 행과 열이 있는 **표 전체** | 엑셀 시트 한 장 |
| **Series** | 1차원 — **열 하나** | 엑셀의 A열 하나 |

````python
print(type(df))
print(type(df["메뉴"]))
````

Series는 4주차의 **리스트와 비슷하지만 더 똑똑한 것**입니다. 리스트는 전체에 2를 곱하려면 for문이 필요했지만, Series는 그냥 곱하면 됩니다.

````python
# 리스트는 이렇게 하면 원하는 결과가 안 나옵니다
수량_리스트 = [10, 20, 30]
print(수량_리스트 * 2)

# Series는 각 값에 곱해집니다
print(df["수량"].head(3) * 2)
````

리스트는 `* 2` 가 **이어붙이기**, Series는 **각 값에 곱하기**였습니다.

이 차이가 Pandas를 쓰는 가장 큰 이유입니다. 수천 행도 반복문 없이 한 줄로 계산됩니다.

1. **총 몇 행인가**
2. **각 열에 값이 몇 개 차 있는가** (`non-null count`) — 빈칸이 있는지 알 수 있습니다
3. **각 열의 자료형이 무엇인가** (`Dtype`)

| Dtype | 뜻 | 4주차 용어로 |
|---|---|---|
| `int64` | 정수 | `int` |
| `float64` | 실수 | `float` |
| `object` 또는 `str` | 글자 | `str` |

**가격 열이 숫자가 아니라 글자로 잡혀 있으면 계산이 안 됩니다.** 4주차에 `"20" + "5"` 가 `205` 가 됐던 그 문제입니다. 실제 업무 데이터에서는 `1,500` 처럼 쉼표가 들어있어서 글자로 읽히는 일이 아주 흔합니다. 그래서 `info()` 를 먼저 봅니다.

### $$$ 점 표기법 다시 보기 — 괄호가 있을 때와 없을 때

지난주 피드백에서 **두 번째로 많이 나온 질문**입니다. 방금 `df.shape` 에는 괄호가 없고 `df.info()` 에는 있었습니다. 왜 다를까요?

4주차에 배운 규칙은 이것이었습니다.

> **괄호가 있으면 "해줘"(메서드), 없으면 "알려줘"(속성)**

이 비유 자체는 잘 전달됐는데, **실제 코드에서 어느 쪽인지 바로 알아보기 어렵다**는 의견이 많았습니다. 그래서 세 가지로 나눠 보겠습니다.

### ① 속성 vs 메서드 — 이미 있는가, 만들어야 하는가

| | 속성 (괄호 없음) | 메서드 (괄호 있음) |
|---|---|---|
| 뜻 | **이미 가지고 있는 정보** | **지금 계산해서 만들어야 하는 것** |
| 예시 | `df.shape`, `df.columns`, `math.pi` | `df.head()`, `df.info()`, `math.sqrt(16)` |

`df.shape` 는 표를 만들 때 이미 정해진 크기입니다. 꺼내 보기만 하면 됩니다.
`df.head()` 는 "앞에서 몇 줄 잘라줘"라는 **작업**입니다. 몇 줄인지도 말해줘야 합니다.

`math.pi` 에 괄호가 없는 이유도 같습니다. 원주율은 **이미 정해진 값**이지 계산할 게 아닙니다. 반면 `math.sqrt(16)` 은 16을 받아서 **지금 계산**해야 합니다.

**구분이 안 될 때는 그냥 둘 다 해보세요.** 틀리면 오류가 알려줍니다.

````python
# 속성에 괄호를 붙이면 어떻게 될까요? 일부러 틀려봅시다
df.shape()
````

마지막 줄: `TypeError: 'tuple' object is not callable`

→ **"이건 시킬 수 있는 게 아닌데요"** 라는 뜻입니다. `callable` 은 "불러서 시킬 수 있는"이라는 뜻입니다.

이 오류가 뜨면 **괄호를 빼면 됩니다.** 반대로 괄호를 빼야 할 걸 안 뺐을 때는 결과가 이상하게 나옵니다.

````python
print(df.shape)       # 괄호 없이: 올바른 결과
print(df.head)        # 괄호 빼먹었을 때: 이상한 게 나옵니다
````

`df.head` 만 쓰면 "이 기능은 여기 있습니다"라는 안내만 나옵니다. **실행은 안 된 것**입니다. 괄호를 붙여야 실제로 작동합니다.

### ② `len()` 은 왜 점을 안 찍나

지난주에 나온 질문입니다. `"안녕".upper()` 는 점을 찍는데 `len("안녕")` 은 왜 안 찍을까요?

| | 모양 | 뜻 |
|---|---|---|
| **함수** | `len(대상)` | 누구에게나 쓸 수 있는 **독립된 도구** |
| **메서드** | `대상.기능()` | **그 대상에만 붙어 있는** 기능 |

`len()` 은 문자열, 리스트, 딕셔너리 어디에나 쓸 수 있는 공용 도구입니다. 그래서 괄호 안에 대상을 넣습니다.
`.upper()` 는 **문자열에만** 있는 기능입니다. 리스트에는 대문자 변환이 없습니다. 그래서 문자열에 붙여서 씁니다.

````python
# 함수: 대상을 괄호 안에 넣는다
print(len("안녕하세요"))
print(len([1, 2, 3]))
print(sum([1, 2, 3]))

# 메서드: 대상에 점을 찍어 붙인다
print("안녕하세요".upper())
print("안녕 하세요".split())
````

### ③ 점이 연달아 나올 때 — `df.isna().sum()`

가장 많이 헷갈린다고 적어준 부분입니다. 읽는 방법은 하나입니다. **왼쪽부터 차례로, 앞의 결과에 다시 시킨다.**

```
df . isna() . sum()
│     │        │
│     │        └─ 그 결과를 세어줘
│     └────────── 빈칸인지 표시해줘
└──────────────── 표에게
```

단계별로 쪼개서 실행하면 바로 보입니다.

````python
# 1단계: 글자 열의 길이를 재면?
단계1 = df["메뉴"].str.len()
print(단계1.head())
````

````python
# 2단계: 그 결과에 다시 평균을 시키면?
단계2 = 단계1.mean()
print(단계2)
````

````python
# 두 단계를 한 줄로 붙인 것이 바로 이것입니다
print(df["메뉴"].str.len().mean())
````

### 3.2. 변수의 종류 — 저장 형식과 의미는 다르다

여기가 오늘 가장 중요한 개념입니다.

`info()` 가 알려준 건 **컴퓨터가 어떻게 저장하고 있는지**입니다. 그런데 분석할 때 필요한 건 **그 값이 무엇을 의미하는지**입니다. 이 둘은 자주 어긋납니다.

통계에서 쓰는 변수 구분은 이렇습니다.

| 종류 | 설명 | 평균을 내도 되나 |
|---|---|---|
| **명목형** (범주형) | 이름일 뿐 순서가 없음 | 안됨 |
| **순서형** | 순서는 있지만 간격이 일정하지 않음 | 조심 |
| **이산형** | 셀 수 있는 숫자 (정수) | 가능 |
| **연속형** | 잴 수 있는 숫자 (소수 가능) | 가능 |

우리 데이터에 적용하면 이렇습니다.

| 열 | 저장 형식 | 변수 종류 | 메모 |
|---|---|---|---|
| `날짜` | 글자 | 시간 | 지금은 글자. 날짜형으로 바꿔야 계산 가능 (6주차) |
| `요일` | 글자 | 명목형 | 월·화·수에 순서가 있다고 볼 수도 있음 |
| `매장`, `메뉴`, `분류`, `결제수단` | 글자 | 명목형 | 평균이 의미 없음 |
| `평점` | **정수** | **순서형** | 숫자지만 조심해야 함 |
| `수량` | 정수 | 이산형 | 반 잔은 팔 수 없음 |
| `가격` | 정수 | 명목형에 가까움 | 메뉴마다 고정값이라 사실상 메뉴의 꼬리표 |
| `후기` | 글자 | **비정형 텍스트** | 표에 담겨 있지만 구조가 없음 |

**`평점` 을 보세요.** `int64` 로 저장돼 있으니 Python은 평균을 계산해 줍니다. 그런데 그 평균이 의미가 있을까요?

별 1개와 2개의 차이가, 별 4개와 5개의 차이와 **같은 크기라고 할 수 있나요?** 사람마다 별점을 주는 기준도 다릅니다. 그래서 평점 평균 3.7은 계산은 되지만 해석에는 주의가 필요합니다.

**`가격` 도 마찬가지입니다.** 숫자지만 메뉴마다 정해진 값이라, 가격 평균은 "메뉴 목록의 평균"이지 "손님이 낸 평균 금액"이 아닙니다. 손님 기준 평균을 구하려면 수량으로 가중해야 합니다.

**컴퓨터는 숫자면 무조건 평균을 내 줍니다. 그게 말이 되는지는 사람이 판단해야합니다.**

````python
# 계산은 됩니다. 하지만 의미가 같을까요?
print("평점 평균:", df["평점"].mean())
print("가격 평균:", df["가격"].mean())
print("수량 평균:", df["수량"].mean())
````

## 6. 정렬하기, 그리고 loc / iloc

`sort_values()` 로 특정 열 기준 정렬을 합니다.

````python
df.sort_values("수량", ascending=False).head()   # 큰 것부터
````

`ascending=False` 가 "내림차순"입니다. 기본값은 `True`(작은 것부터)입니다.

> **주의**: `sort_values()` 는 정렬된 **사본**을 돌려줄 뿐 원본 `df` 는 그대로입니다. 결과를 계속 쓰려면 변수에 담아야 합니다. Pandas 메서드는 대부분 이렇습니다. "실행했는데 왜 안 바뀌었지" 싶을 때 이걸 의심하세요.

````python
정렬본 = df.sort_values("수량", ascending=False)

print(정렬본.head(3)["수량"].tolist())    # 정렬됨
print(df.head(3)["수량"].tolist())        # 원본은 그대로
````

### loc 와 iloc — 읽을 수 있을 정도만

| | 기준 | 예시 |
|---|---|---|
| `iloc` | **순서 번호** | `df.iloc[0]` → 첫 번째 행 |
| `loc` | **이름표** | `df.loc[0, "메뉴"]` → 0번 행의 메뉴 |

실제로는 조건 필터링(5절)을 훨씬 많이 씁니다. 지금은 눈에 익히는 정도면 충분합니다.

````python
print(df.iloc[0]["메뉴"])        # 첫 번째 행의 메뉴
print(df.loc[0, "메뉴"])         # 같은 결과
print(df.iloc[0:3].shape)        # 0~2번 (끝 제외)
print(df.loc[0:3].shape)         # 0~3번 (끝 포함!)
````

## 8. 글자가 담긴 열 다루기

여러분이 앞으로 만날 데이터에는 **숫자가 아니라 글자**가 들어 있는 경우가 많습니다. 설문의 자유응답, 상품 리뷰, 기사 본문, 인터뷰 전사본 같은 것들입니다.

이런 걸 **비정형 데이터**라고 합니다. 행과 열로 딱 떨어지지 않는 자료입니다.

좋은 소식은, **표의 한 열에 담기기만 하면 지금까지 배운 방법을 거의 그대로 쓸 수 있다**는 것입니다. 우리 데이터의 `후기` 열로 해보겠습니다.

````python
df["후기"].head(10)
````

### .str 액세서라 — 글자 열 전용 도구

글자 열에는 `.str` 을 붙이면 글자 전용 기능들이 열립니다. **4주차에 배운 문자열 메서드와 이름이 같습니다.**

| 4주차에 배운 것 | Pandas에서 |
|---|---|
| `"안녕".replace("안", "반")` | `df["후기"].str.replace("안", "반")` |
| `"안녕 하세요".split()` | `df["후기"].str.split()` |
| `len("안녕")` | `df["후기"].str.len()` |

차이는 하나입니다. 4주차엔 **글자 하나**에 적용했고, 지금은 **열 전체(90개)** 에 한 번에 적용됩니다.

````python
# 후기 길이 재기
df["후기길이"] = df["후기"].str.len()

df[["후기", "후기길이"]].head()
````

````python
print("가장 긴 후기 길이:", df["후기길이"].max())
print("평균 후기 길이:", df["후기길이"].mean())
````

### str.contains(): 특정 단어가 들어있는가

**글자 열에서 가장 많이 쓰는 기능**입니다. 조건 필터링(5절)과 똑같이 쓰면 됩니다.

````python
df["후기"].str.contains("대기").head(10)
````

5절에서 본 True/False가 또 나왔습니다. 그러면 대괄호에 넣으면 되겠죠.

````python
df[df["후기"].str.contains("대기")]
````

`"대기"` 가 들어간 후기만 골라냈습니다. **질적 자료에서 특정 주제가 언급된 사례를 찾는 작업**이 바로 이겁니다.

몇 건인지, 전체의 몇 퍼센트인지도 바로 셀 수 있습니다.

````python
대기언급 = df[df["후기"].str.contains("대기")]

print("대기 언급 건수:", len(대기언급))
print("전체 대비 비율:", len(대기언급) / len(df) * 100, "%")
print("대기 언급 후기의 평균 평점:", 대기언급["평점"].mean())
print("전체 평균 평점:", df["평점"].mean())
````

마지막 두 줄을 비교해 보세요. 대기를 언급한 후기의 평점이 전체보다 낮다면, **대기 시간이 불만의 원인일 가능성**을 시사합니다.

다만 **시사할 뿐 증명하지는 않습니다.** 짚어야 할 게 두 가지입니다.

첫째, **건수가 너무 적습니다.** 몇 건밖에 안 되는 사례로 평균을 내면 한두 개가 통째로 결과를 흔듭니다. "차이가 보인다"와 "차이가 있다"는 다른 말입니다. 이 간극을 다루는 방법은 통계학 과목에서 배우고, 이 수업에서는 **앞의 표현까지만** 씁니다.

둘째, **방향을 알 수 없습니다.** 대기가 길어서 평점이 낮은 건지, 원래 불만인 사람이 이유를 찾다가 대기를 언급한 건지는 이 데이터로 구분되지 않습니다.

### 키워드를 여러 개 세어보기

````python
키워드들 = ["맛", "대기", "친절", "가격", "자리"]

for 키워드 in 키워드들:
    건수 = df["후기"].str.contains(키워드).sum()
    print(f"{키워드}: {건수}건")
````

4주차에 배운 for문이 여기서 쓰였습니다. `True` 는 1, `False` 는 0으로 취급되어서 `.sum()` 이 곧 "참인 개수"가 됩니다.

### 단어 수 세기

````python
df["단어수"] = df["후기"].str.split().str.len()

df[["후기", "단어수"]].head()
````

`split()` 으로 쪼갠 뒤 `len()` 으로 개수를 셌습니다. 점이 연달아 나오는 구조인데, **왼쪽부터 차례로** 읽으면 됩니다(4주차).

>
> - `"맛없어요"` 와 `"맛있어요"` 가 **둘 다 "맛" 언급**으로 세어집니다

> - 반어법이나 비꼬는 표현은 전혀 잡히지 않습니다
